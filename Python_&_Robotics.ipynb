```python
import math
import matplotlib.pyplot as plt


print("\nEnter the three sides of a triangle:")

a = float(input("Side 1: "))
b = float(input("Side 2: "))
c = float(input("Side 3: "))

# Step 2: Check triangle validity
if (a + b > c) and (a + c > b) and (b + c > a):
    print("Triangle is valid")

    # Step 3: Type of triangle


    if a == b and b == c:
        print("Equilateral Triangle")
    elif a == b or b == c or a == c:
        print("Isosceles Triangle")
    else:
        print("Scalene Triangle")

    # Right-angled check
    if (a*a + b*b == c*c) or (a*a + c*c == b*b) or (b*b + c*c == a*a):
        print("It is also a Right-Angled Triangle")

    # Step 4: Area and Perimeter
    s = (a + b + c) / 2
    area = math.sqrt(s * (s - a) * (s - b) * (s - c))
    perimeter = a + b + c

    print("Perimeter =", perimeter)
    print("Area =", area)

```

```python
SETS = {}  # dictionary to store sets

# ---------------------------------------------
# Function: Create a Set
# ---------------------------------------------
def create_set():
    name = input("Enter set name (e.g., A, B): ").strip()
    elements = input("Enter elements (comma-separated): ")

    # convert to set, remove spaces
    element_set = ([e.strip() for e in elements.split(",") if e.strip() != ""])

    SETS[name] = element_set
    print(f"Set {name} created: {element_set}\n")

create_set()
```
